```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

def test_setup():
    print("--- Starting Environment Test ---")
    
    # 1. Test Data Fetching (yfinance)
    ticker = "GC=F" # Gold Futures
    print(f"Fetching data for {ticker}...")
    data = yf.download(ticker, period="60d", interval="1d")
    
    if data.empty:
        print("❌ Error: Could not fetch data. Check your internet connection.")
        return

    # 2. Test Data Manipulation (pandas)
    # Calculating a simple 7-day moving average
    data['SMA_7'] = data['Close'].rolling(window=7).mean()
    print("✅ Data downloaded and processed successfully.")

    # 3. Test Visualization (matplotlib)
    print("Generating plot...")
    plt.figure(figsize=(10, 5))
    plt.plot(data['Close'], label='Gold Close Price', color='gold')
    plt.plot(data['SMA_7'], label='7-Day Moving Avg', color='black', linestyle='--')
    
    plt.title(f"{ticker} Price Test")
    plt.xlabel("Date")
    plt.ylabel("Price (USD)")
    plt.legend()
    plt.grid(True)
    
    print("✅ Plotting engine working. (Close the window to finish)")
    plt.show()

if __name__ == "__main__":
    test_setup()
```
